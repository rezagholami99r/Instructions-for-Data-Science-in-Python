```python
# SEMI join in 3 steps: 
# "Which rows from the LEFT table have a matching key in the RIGHT table?"

# 1. Doing an inner join between 2 tables
inner_join_table = LEFT_table.merge(RIGHT_table, on='C')

# 2. Searching for key columnon the LEFT table, using ".isin()" to create a boolean series
LEFT_table['C'].isin(inner_join_table['C'])

# 3. Subsetting the rows of the left table
SEMI_JOIN = LEFT_table[LEFT_table['C'].isin(inner_join_table['C'])] 
```

![My Chart](../images/8.png)

```python
# ANTI join in 3 steps:

# 1. Doing a left join between 2 tables and adding an indicator column
# NOTE: You must USE "indicator = True" argument to have a last column in left join to illustrate each row belongs to what table. (2 possible contents: "left_only", "both")
left_join_table = LEFT_table.merge(RIGHT_table, on='C', how='left', indicator=True)

# 2. Searching for rows that exist only in the LEFT table
unmatched_rows = left_join_table.loc[left_join_table['_merge'] == 'left_only', 'C']

# 3. Subsetting the rows of the LEFT table
anti_join_table = LEFT_table[LEFT_table['C'].isin(unmatched_rows)]
```
