### **Default arguments, variable-length arguments and scope**

```python
# Scope: a part of the program where an object or name may be accessible
#        1. Global scope - defined in the main body of a script
#        2. Local scope - defined inside a function
#        3. Built-in scope - names in the pre-defined built-ins module

num = 5
def func1():
    num = 3
    print(num)

def func2():
    global num
    double_num = num * 2
    num = 6
    print(double_num)          # Important: AFTER calling func1() and func2() the num is 6
# NOTE: By defining "global" in the function, the variable(s) follow(s) will be the global

import builtins                # To see all built-in scopes. SEE THEM!
dir(builtins)
------------------------------------------------------------------------------------------
def outer():                   # Nested Function
    x = 10
    def inner():
        print(x)
    inner()
outer()
# NOTE: Think of nested functions like a person living inside a house: inner() can see x because x belongs to the surrounding outer() function.
# NOTE2: You can use a variable in front of "nonlocal" scope inside the inner function.
# NOTE3: "nonlocal" says: "I want to modify a variable from the nearest surrounding function." : "I want to modify a variable from the global/module scope."
------------------------------------------------------------------------------------------
# FLEXIBLE FUNCTIONS
def powerrr(number, pow=1)    # "pow" argument is a default and flexible argument
    """Raise number to the power of pow.""" 
    new_value = number ** pow 
    return new_value

def add_allll(*args):         # You can call multiple arguments "as you want" in ()
    """Sum all values in *args together.""" 
    # Initialize sum 
    sum_all = 0 
    # Accumulate the sum 
    for num in args: 
        sum_all += num 
    return sum_all

def print_allll(**kwargs):    # Exactly the same but for dictionaries with key-value pairs
    """Print out key-value pairs in **kwargs.""" 
    # Print out the key-value pairs 
    for key, value in kwargs.items(): 
        print(key + ": " + value) 

```
