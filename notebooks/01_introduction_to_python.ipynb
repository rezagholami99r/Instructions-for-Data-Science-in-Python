### **ADD COLUMNS WITH "FOR"**

```python
# example 1: adding column to show the length of "country" column digits (in Pandas)
for lab, row in brics.iterrows():
    brics.loc[lab, "name_length"] = len(row["country"])

# example 2: adding column to show the upper case of "country" strings (in Pandas)
for labe, row in cars.iterrows():
    cars.loc[labe, "COUNTRY"] = row["country"].upper()


# apply() : efficient way to add columns INSTEAD OF .iterrows() Method
# the iterrows() method in combination with a for loop is not the preferred way  to go. Instead, you'll want to use apply(). (same examples)
for lab, row in brics.iterrows() :
    brics["name_length"] = brics["country"].apply(len)

for lab, row in cars.iterrows() :
    cars["COUNTRY"] = cars["country"].apply(str.upper)
```

```python
np.random.seed()  
# As an argument, it takes an integer of your choosing. If you call the function, no output will be generated.

np.random.rand()
# if you don't specify any arguments, it generates a random float between 0 and 1.

np.random.randint(a, b)                         # Generate random integer between a to b-1
------------------------------------------------------------------------------------------
# NumPy is imported, seed is set                                   NICE EXAMPLE

# Initialize random_walk (you start with number 0 : not empty[])
random_walk = [0]

for x in range(100) :
    step = random_walk[-1]                           # my_list[-1] shows the LAST element
    dice = np.random.randint(1,7)

    if dice <= 2:
        # Replace below: use max to make sure step can't go below 0
        step = max(0, step - 1)
    elif dice <= 5:
        step = step + 1
    else:
        step = step + np.random.randint(1,7)

    random_walk.append(step)

print(random_walk)
------------------------------------------------------------------------------------------
all_walks = []                                   # COMPLETE example of above: Distribution
for i in range(500) :
    random_walk = [0]
    for x in range(100) :
        step = random_walk[-1]
        dice = np.random.randint(1,7)
        if dice <= 2:
            step = max(0, step - 1)
        elif dice <= 5:
            step = step + 1
        else:
            step = step + np.random.randint(1,7)
        if np.random.rand() <= 0.001 :
            step = 0
        random_walk.append(step)
    all_walks.append(random_walk)

# Create and plot np_aw_t
np_aw_t = np.transpose(np.array(all_walks))

# Select last row from np_aw_t: ends
ends = np_aw_t[-1,:]

plt.hist(ends)                            # this histogram showcases a Normal distribution
plt.show()
# NOTE: The histogram of the previous exercise was created from a NumPy array ends, that contains 500 integers. Each integer represents the end point of a random walk. To calculate the chance that this end point is greater than or equal to 60, you can count the number of integers in ends that are greater than or equal to 60 and divide that number by 500, the total number of simulations.
np.sum(ends >= 60)/ 500
```
