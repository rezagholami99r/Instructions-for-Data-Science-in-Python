### **List comprehensions and generators**

```python
# List comprehensions: Collapse for loops to build lists into a single line (Components: Iterable, Iterator variable (represent members of iterable))

# Examples of List comprehensions (Tradeoff: readability!):
new_nums = [num + 1 for num in nums]
result = [num for num in range(11)] 
pairs_2 = [(num1, num2) for num1 in range(0, 2) for num2 in range(6, 8)]

# NOTE: Matrices can be represented as a list of lists in Python.

# Create a 5 x 5 matrix using a list of lists: matrix
matrix = [[col for col in range(5)] for row in range(5)]
# Print the matrix
for row in matrix:
    print(row)

[num ** 2 for num in range(10) if num % 2 == 0]          # Conditional List Comprehensions
[num ** 2 if num % 2 == 0 else 0 for num in range(10)]   # With if-else
# NOTE: Notice the ORDER difference between "if" L.C. and the "if-else" one!!!!!!!!!!

{num: -num for num in range(9)}                          # Dict List Comprehensions
------------------------------------------------------------------------------------------
# Generator: is exactly the same as the List Comprehension but instead "[]", using "()".

# List comprehension: returns a "list or Dictionary". (using memory)
# Generators: returns a "generator object" (NOT using memory: used for HEAVY calculations)
even_nums = (num for num in range(10) if num % 2 == 0)     # generator
print(list(even_nums))                                     # printing generator as a list

# Build a generator function (USING yeild): Below is the example!
def num_sequence(n): 
    """Generate values from 0 to n.""" 
    i = 0 
    while i < n: 
        yield i                       # used when you want to build a Generator functions
        i += 1

list = ['cersei', 'jaime', 'tywin', 'tyrion', 'joffrey']      # example1
lengths = (len(person) for person in list)
for value in lengths:
    print(value)

def get_lengths(input_list):                                  # example2 (generator func.)
    """Generator function that yields the
    length of the strings in input_list."""
    for person in input_list:
        yield len(person)

for value in get_lengths(lannister):
    print(value)
```
