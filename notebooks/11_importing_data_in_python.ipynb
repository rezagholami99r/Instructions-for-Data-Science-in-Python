### **Introduction and flat files**

```python
file = open('forexample.txt', mode='r')     # 'r' is to read, 'w' is to write 
text = file.read()                          # or .readline()
file.close()

with open('forexample.txt', 'r') as file:   # another way (Context manager with)
    print(file.read())

# You can print the Zen of Python in your shell by typing "import this" into it! 
# Flat files: Text files containing records. That is, table data. (NOT relational ones)
#             (Record: row of fields or attributes, Column: feature or attribute)

import numpy as np 
filename = 'forexample.txt'                 # Importing with numpy 

data = np.loadtxt(filename, delimiter=',', skiprows=1, usecols=[0, 2], dtype=...)

# delimiter: ',' for comma-delimited, '\t' for tab-delimited.
# skiprows: allows you to specify how many rows (not indices) you wish to skip.
# usecols takes a list of the indices of the columns you wish to keep.
------------------------------------------------------------------------------------------
import pandas as pd              
filename = 'forexample.csv'                 # Importing with pandas
data = pd.read_csv(filename, nrows =, header =, sep='', comment='', na_values=[]) 
# nrows: specifies how many rows to read from the file

# header: accepts row numbers to use as the column labels and marks the start of the data. If the file does not contain a header row, you can set header=None, and pandas will automatically assign integer column labels starting from 0 (e.g., 0, 1, 2, …).

# sep: sets the expected delimiter: ',' for comma-delimited, '\t' for tab-delimited.

# comment: takes characters that comments occur after in the file, indicating that any text starting with these characters should be ignored.

# na_values: takes a list of strings to identify as NA/NaN. By default, some values are already recognized as NA/NaN. Providing this argument will supply additional values.

data_array = data.to_numpy()                # Converting to numpy
```
