### **Visualizing Two Quantitative Variables**

```python
# Using .relplot() instead: (filtering through the creating more plots)
sns.relplot(x="var1", y="var2", data=DF, kind="scatter", col="var3", row="var4", col_wrap=      
            , col_order=["", "", "", ...]) )  
# similarly for row_wrap and row_order arguments

sns.relplot(x, y, data, kind, hue, size="var", style="var", alpha)   # alpha: transparency

# NOTE: To create a line plot, everything is similar to the scatter but: kind = "line"
# NOTE2: For replacing confidence interval with standard deviation: errorbar = "sd". The default is errorbar = None.
sns.relplot(x, y, data, kind="line", hue, size, style, markers, dashes, errorbar)
```

### **Visualizing a Categorical and a Quantitative Variable**

```python
sns.countplot(x = "var", data = DF)                                            # countplot
category_order = ["extremely low", "low", "moderate", "high", "extremely high"]# Example
sns.catplot(x="var", data=DF, kind="count", order=category_order)

# Note: By using "y" instead of "x", you make the chart horizontal (not vertical).
# NOTE2: All of the arguments used in relplot() can deploy here.
# NOTE3: You can have bar plot by kind = "bar". In this plot you have a 95% confidence intervals for the mean.(You can set errorbar = None to do not see the confidence interval)
------------------------------------------------------------------------------------------
sns.catplot(x="var1", y="var2", data=DF, kind="box", showfliers=False) 
sns.catplot(x="var1", y="var2", data=DF, kind="box", whis=[m, n])
# Notes about boxplot's whis: 1. By default, the whiskers extend to (1.5 * irq) 
#                             2. Make them extend to (2.0 * IQR): whis=2.0
#                             3. Show the 5th and 95th percentiles: whis=[5, 95]
#                             4. Show min and max values: whis=[0, 100]
------------------------------------------------------------------------------------------
# Point Plot: Points show mean of quantitative variable. Vertical lines show 95% confidence intervals. Line plot has quantitative variable (usually time) on x-axis BUT Point plot has categorical variable x-axis.(errorbar=None: REVOVING confidence Interval)

from numpy import median                          # Displaying the median instead of mean
sns.catplot(x="var1", y="var2", data=DF, kind="point", estimator=median)
# Note : argument linestyle="none" to removing the line.
```
